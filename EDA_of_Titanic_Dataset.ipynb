```python
df=pd.read_excel(r"/content/Titanic_dataset.xlsx")
```

```python
print(df.info())
print(df.describe())
```

```python
#UNIVARIATE ANALYSIS

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Survival Distribution (0 = Died, 1 = Survived)
sns.countplot(x='Survived', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Overall Survival Count')

# Passenger Class Distribution
sns.countplot(x='Pclass', data=df, ax=axes[1], palette='muted')
axes[1].set_title('Passenger Count by Class')

# Age Distribution
sns.histplot(df['Age'], bins=20, kde=True, ax=axes[2], color='skyblue')
axes[2].set_title('Age Distribution of Passengers')

plt.tight_layout()
plt.show()
```

```python
#BIVARIATE ANALYSIS
#A)SURVIVAL BY GENDER

plt.figure(figsize=(6, 4))
sns.countplot(x='Sex', hue='Survived', data=df, palette='Set1')
plt.title('Survival Count by Gender')
plt.show()
```

```python
#B)SURVIVAL BY PASSENGER CLASS

plt.figure(figsize=(6, 4))
sns.countplot(x='Pclass', hue='Survived', data=df, palette='viridis')
plt.title('Survival Count by Passenger Class')
plt.show()
```

```python
#C)SURVIVAL BY AGE DISTRIBUTION

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', hue='Survived', multiple='stack', bins=30, palette='coolwarm')
plt.title('Age Distribution Split by Survival Status')
plt.xlabel('Age')
plt.show()
```

```python
#MULTIVARIATE ANALYSIS

sns.catplot(x='Pclass', y='Survived', hue='Sex', kind='point', data=df, palette='dark')
plt.title('Survival Rate across Class and Gender')
plt.show()
```
